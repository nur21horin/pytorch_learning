```python
#fp
import numpy as np

# Step 1: Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 2: Inputs
x1 = 0.6


# Step 3: Hidden Layer Weights and Bias
w1 = 0.4
w2 = 0.7
b1 = 0.2

# Step 4: Hidden Neuron Calculation
z1 = (x1 * w1)  + b1
h1 = sigmoid(z1)

# Step 5: Output Layer Weight and Bias
w3 = 0.9
b2 = 0.1

# Step 6: Output Neuron Calculation
z2 = (h1 * w2)
h2 = sigmoid(z2)

z3= (h2 * w3) + b2
h3 = sigmoid(z3)

# Step 7: Actual Output
t1 = 1

# Step 8: Error and Loss
error = t1 - h3
loss = 0.5 * (error ** 2)

# Display Results
print("===== FORWARD PROPAGATION =====")
print("Hidden Net Input (z1):", z1)
print("Hidden Output (h1):", h1)

print("\nHidden Net Input (z2):", z2)
print("Hidden Output (h2):", h2)

print("\nOutput Net Input (z3):", z3)
print("Predicted Output (h3):", h3)

print("\nActual Output (t1):", t1)
print("Error:", error)
print("Loss:", loss)





```

```python

def sigmoid_derivative(x):
    return x * (1 - x)

# Learning Rate
lr = 0.1

#bp

delta_h3 = error * sigmoid_derivative(h3)

# 2. Hidden Layer 2  h2 - z2
delta_h2 = delta_h3 * w3 * sigmoid_derivative(h2)

# 3. Hidden layer - z1

delta_h1 = delta_h2 * w2 * sigmoid_derivative(h1)



# Update Output Layer (w3, b2)
w3_new = w3 + (lr * h2 * delta_h3)
b2_new = b2 + (lr * delta_h3)

# Update Hidden Layer 2 (w2)
w2_new = w2 + (lr * h1 * delta_h2)

# Update Hidden Layer 1 (w1, b1)
w1_new = w1 + (lr * x1 * delta_h1)
b1_new = b1 + (lr * delta_h1)

# Results
print(" Backprropagation: ")
print(f"delta_h3 (Output Gradient): {delta_h3:.6f}")
print(f"delta_h2 (Hidden 2 Gradient): {delta_h2:.6f}")
print(f"delta_h1 (Hidden 1 Gradient): {delta_h1:.6f}")

print(f"w3: {w3:.6f} -> {w3_new:.6f}")
print(f"b2: {b2:.6f} -> {b2_new:.6f}")
print(f"w2: {w2:.6f} -> {w2_new:.6f}")
print(f"w1: {w1:.6f} -> {w1_new:.6f}")
print(f"b1: {b1:.6f} -> {b1_new:.6f}")
```

```python
# Training Loop for the new 3-layer architecture
for epoch in range(500):

    # Forward Propagation
    z1 = (x1 * w1) + b1
    h1 = sigmoid(z1)

    z2 = (h1 * w2)
    h2 = sigmoid(z2)

    z3 = (h2 * w3) + b2
    h3 = sigmoid(z3)

    # Error and Loss
    error = t1 - h3
    loss = 0.5 * (error ** 2)

    # Backpropagation
    delta_h3 = error * sigmoid_derivative(h3)
    delta_h2 = delta_h3 * w3 * sigmoid_derivative(h2)
    delta_h1 = delta_h2 * w2 * sigmoid_derivative(h1)

    # Weight Updates
    w3 = w3 + (lr * h2 * delta_h3)
    b2 = b2 + (lr * delta_h3)

    w2 = w2 + (lr * h1 * delta_h2)

    w1 = w1 + (lr * x1 * delta_h1)
    b1 = b1 + (lr * delta_h1)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Output = {h3:.4f}, Error = {error:.4f}, Loss = {loss:.4f}")

print("\Final result")
print("Final Output =", h3)
print("Final Error =", error)
print("Final Loss =", loss)
```

```python
#matrix operation
f=torch.randint(size=(2,3),low=0,high=10,dtype=torch.float32)
g=torch.randint(size=(3,3),low=0,high=10,dtype=torch.float32)
```

```python
print(f)
print(g)
```

```python
torch.matmul(f,g)
```
